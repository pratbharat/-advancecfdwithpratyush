```python
import pyvista as pv

case =  pv.OpenFOAMReader(r"\\wsl.localhost\Ubuntu-20.04\home\iterasim\OpenFOAM\iterasim-9\run\pitzDaily\test.foam")
zmin , zmax = mesh1.bounds[4] , mesh1.bounds[5]
slice_origin = [
    (mesh.bounds[0] + mesh.bounds[1])/2, 
    (mesh.bounds[1] + mesh.bounds[2])/2, 
    (zmin + zmax)/2
]
time = case.time_values
for t in time:
    case.set_active_time_value(t)
    mesh1 = case.read() 
    slice_mesh = mesh1.slice(normal= [0, 0, 1], origin = slice_origin )
    p2 = pv.Plotter()
    p2.add_mesh(slice_mesh, scalars='U', cmap= 'jet')
    p2.view_xy()
    filename = f"slice_p_t{t:.4f}.png"       # filename per time
    p2.screenshot(filename)
    p2.close()
```
